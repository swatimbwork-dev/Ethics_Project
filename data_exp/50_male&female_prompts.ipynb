```python
import pandas as pd


# ==========================================
# 1. Read Sofia's final dataset
# ==========================================

df = pd.read_csv("final_data_with_prompts.csv")

print("Original dataset size:", len(df))


# ==========================================
# 2. Remove unnecessary Unnamed column
# ==========================================

df = df.loc[:, ~df.columns.str.contains("^Unnamed")]


# ==========================================
# 3. Select Male patients
# ==========================================
# Conditions:
# Gender = Male
# Age = [40-50)
# Race = AfricanAmerican or Caucasian
# Pain levels = 0, 2, 4, 6, 8

male_selected = df[
    (df["gender"] == "Male") &
    (df["age"] == "[40-50)") &
    (df["race"].isin([
        "AfricanAmerican",
        "Caucasian"
    ])) &
    (df["pain_level"].isin([
        0, 2, 4, 6, 8
    ]))
].copy()


print("\nAvailable patients after filtering:")
print(
    pd.crosstab(
        male_selected["pain_level"],
        male_selected["race"]
    )
)


# ==========================================
# 4. Randomly select 5 patients
#    from each Pain Level × Race group
# ==========================================

male50 = (
    male_selected
    .groupby(
        ["pain_level", "race"],
        group_keys=False
    )
    .sample(
        n=5,
        random_state=42
    )
    .sort_values(
        ["pain_level", "race"]
    )
    .reset_index(drop=True)
)


# ==========================================
# 5. Keep only necessary columns
# ==========================================

male50 = male50[
    [
        "race",
        "gender",
        "age",
        "pain_level",
        "prompt"
    ]
]


# ==========================================
# 6. Check the selected data
# ==========================================

print("\nTotal selected:", len(male50))

print("\nPain level × Race:")
print(
    pd.crosstab(
        male50["pain_level"],
        male50["race"]
    )
)

print("\nGender:")
print(
    male50["gender"].value_counts()
)

print("\nAge:")
print(
    male50["age"].value_counts()
)


# ==========================================
# 7. Save male 50 prompts
# ==========================================

male50.to_csv(
    "male_50_prompts.csv",
    index=False
)

print("\nSaved successfully!")
print("File: male_50_prompts.csv")


# ==========================================
# 8. Display final dataset
# ==========================================

male50
```

```python
# ==========================================
# Select 50 Female patients
# ==========================================

female_selected = df[
    (df["gender"] == "Female") &
    (df["age"] == "[40-50)") &
    (df["race"].isin(["AfricanAmerican", "Caucasian"])) &
    (df["pain_level"].isin([0, 2, 4, 6, 8]))
]


# 5 patients for each Pain Level × Race combination
female50 = (
    female_selected
    .groupby(["pain_level", "race"])
    .sample(n=5, random_state=42)
    .sort_values(["pain_level", "race"])
    .reset_index(drop=True)
)


# Keep the columns we need
female50 = female50[
    [
        "race",
        "gender",
        "age",
        "pain_level",
        "prompt"
    ]
]


# Save the 50 prompts
female50.to_csv(
    "female_50_prompts.csv",
    index=False
)


# Check the result
print("Total:", len(female50))

print("\nPain level × Race:")
print(
    pd.crosstab(
        female50["pain_level"],
        female50["race"]
    )
)

female50
```
